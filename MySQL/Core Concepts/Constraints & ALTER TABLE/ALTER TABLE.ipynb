# ALTER TABLE

## ADD a column
 ```sql
ALTER TABLE table_name
ADD COLUMN column_name VARCHAR(100);
```

````python
%%sql

USE dog_world;

SELECT *
FROM users
````

````python
%%sql

ALTER TABLE users
ADD COLUMN user_type VARCHAR(50) NOT NULL;
````

````python
%%sql

INSERT INTO users (name, age, user_type)
VALUES ('liam', 43, 'team')
````

````python
%%sql

SELECT *
FROM users
````

## DROP a column
 ```sql
ALTER TABLE table_name
DROP COLUMN column_name;
```

````python
%%sql

ALTER TABLE users
DROP COLUMN user_type;
````

````python
%%sql

SELECT *
FROM users
````

## Renaming Tables
 ```sql
RENAME TABLE orig_table_name TO new_table_name
```
#### Other way
 ```sql
ALTER TABLE orig_table_name RENAME TO new_table_name
```

````python
%%sql

ALTER TABLE users RENAME TO existing_users
````

## Renaming Columns
```sql
ALTER TABLE table_name
RENAME COLUMN orig_col_name TO new_col_name
```

````python
%%sql

ALTER TABLE existing_users
RENAME COLUMN name TO existing_names
````

````python
%%sql

SELECT *
FROM existing_users
````

## Modify Columns
##### By modifying columns we mean that we are actually modifying the Definition of the column
```sql
ALTER TABLE table_name
MODIFY column_name VARCHAR(100)
```

## Changing Columns
#### We can use CHANGE to rename a column AND change its datatype
```sql
ALTER TABLE table_name
CHANGE org_col_name new_col_name VARCHAR(50)
```

````python
%%sql

DESCRIBE existing_users
````

````python
%%sql

ALTER TABLE existing_users
MODIFY existing_names VARCHAR(200) NOT NULL;
````

````python
%%sql

DESCRIBE existing_users
````

#  ALTER TABLE CONSTRAINTS
## ADD or DROP Constraints once the table is created
```sql
ALTER TABLE table_name
ADD CONSTRAINT constraint_name CHECK (expression)
```

#### DROP
```sql
ALTER TABLE table_name
DROP CONSTRAINT constraint_name
```

````python
%%sql

ALTER TABLE houses
ADD CONSTRAINT equal_pricing CHECK (sale_price > purchase_price)
````

````python
%%sql

ALTER TABLE houses
DROP CONSTRAINT equal_pricing
````
