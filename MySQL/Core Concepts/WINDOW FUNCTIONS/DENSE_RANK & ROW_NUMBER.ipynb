# DENSE_RANK & ROW_NUMBER()

# ROW_NUMBER()
#### ROW_NUMBER() window function assigns a unique, sequential integer to each row in a result set, starting at 1 for the first row.

```python
%%sql
USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    ROW_NUMBER() OVER(PARTITION BY department ORDER BY salary DESC) as dept_row_number,
    RANK() OVER(ORDER BY salary DESC) AS overall_salary_rank
FROM employees
ORDER BY department
```

# DENSE RANK
#### DENSE_RANK() window function in MySQL assigns a sequential rank to each row within a partition, ensuring that tied values receive the same rank and no ranking numbers are skipped.

```python
%%sql
USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    ROW_NUMBER() OVER(PARTITION BY department ORDER BY salary DESC) as dept_row_number,
    RANK() OVER(ORDER BY salary DESC) AS overall_salary_rank,
    DENSE_RANK() OVER(ORDER BY salary DESC) AS overall_dense_rank
FROM employees
ORDER BY overall_dense_rank
```
