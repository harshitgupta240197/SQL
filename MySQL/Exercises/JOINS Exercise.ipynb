# JOINS Exercise

```python
%%sql

USE dog_world;

CREATE TABLE students
    (
        id INT AUTO_INCREMENT PRIMARY KEY,
        first_name VARCHAR(50)
    )
```

```python
%%sql

CREATE TABLE papers
    (
        title VARCHAR(50),
        grade INT,
        student_id INT,
        FOREIGN KEY (student_id) REFERENCES students(id)
    )
```

```python
%%sql

SELECT 
    s.first_name, 
    p.title,
    p.grade
FROM students as s 
JOIN papers as p 
ON s.id = p.student_id
ORDER BY grade DESC
```

```python
%%sql

SELECT 
    s.first_name, 
    p.title,
    p.grade
FROM students as s 
LEFT JOIN papers as p 
ON s.id = p.student_id
```

```python
%%sql

SELECT 
    s.first_name, 
    IFNULL(p.title, 'MISSING'),
    IFNULL(p.grade, 0)
FROM students as s 
LEFT JOIN papers as p 
ON s.id = p.student_id
```

```python
%%sql

SELECT
    s.first_name,
    IFNULL(AVG(grade), 0) as average
FROM students as s 
LEFT JOIN papers as p 
ON s.id = p.student_id
GROUP BY s.first_name
ORDER BY average DESC
```

```python
%%sql

SELECT
    s.first_name,
    IFNULL(AVG(grade), 0) as average,
    CASE 
        WHEN IFNULL(AVG(grade), 0) > 75 THEN 'PASSING'
        ELSE 'FAILING'
    END as passing_status
FROM students as s 
LEFT JOIN papers as p 
ON s.id = p.student_id
GROUP BY s.first_name
ORDER BY average DESC
```
