# REVERSE - String Function
#### It will take in whatever string we provide it and it will REVERESE it
```sql
SELECT 
    REVERSE(column_name)
FROM table_name
```

````python
%%sql

USE book_shop;

SELECT 
    REVERSE(title)
FROM books
LIMIT 10;
````

````python
%%sql

SELECT 
    CONCAT(
        author_fname, 
        REVERSE(author_fname)
        ) AS weird_string
FROM books
LIMIT 10;
````
