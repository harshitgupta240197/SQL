# MIN & MAX

```python
%%sql

USE book_shop;

SELECT 
    MIN(released_year) AS Min_Year
FROM books
```

```python
%%sql

USE book_shop;

SELECT 
    MAX(released_year) AS Max_Year
FROM books
```

```python
%%sql

SELECT 
    MIN(author_lname)
FROM books
```

```python
%%sql

SELECT 
    MAX(author_lname)
FROM books
```

#### I want to know the title of the longest book

```python
%%sql

SELECT
    title,
    MAX(pages) AS Max_Pages
FROM books
GROUP BY title
ORDER BY Max_Pages DESC
LIMIT 1;
```
