# CHAR_LENGTH - String Function
### CHAR_LENGTH() counts characters, while LENGTH() counts bytes
```sql
SELECT 
    CHAR_LENGTH(column_name)
FROM table_name
```

````python
%%sql

USE book_shop;

SELECT *
FROM books
LIMIT 5;
````

````python
%%sql

SELECT
    CHAR_LENGTH(title) AS title_length,
    author_fname,
    author_lname
FROM books
LIMIT 5;
````
