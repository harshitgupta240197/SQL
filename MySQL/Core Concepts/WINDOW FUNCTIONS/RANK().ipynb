# RANK()
#### RANK() window function in MySQL assigns a rank to each row within a partition of a result set, producing non-consecutive ranks with gaps when identical values (ties) are found.

#### From the employees data use RANK() to find out the ranking by salary

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    RANK() OVER(ORDER BY salary DESC) AS overall_salary_rank
FROM employees
```

#### Find the department salary rank

```python
%%sql

SELECT 
    emp_no, 
    department,
    salary,
    RANK() OVER(PARTITION BY department ORDER BY salary DESC) AS dept_salary_rank,
    RANK() OVER(ORDER BY salary DESC) AS overall_salary_rank
FROM employees
ORDER BY department
```
