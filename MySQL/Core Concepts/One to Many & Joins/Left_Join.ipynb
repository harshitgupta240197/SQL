# LEFT JOIN
#### SQL LEFT JOIN (or LEFT OUTER JOIN) returns all records from the left table and the matching records from the right table. If there is no match for a row, the result will contain NULL values for all columns of the right table.

````python
%%sql

USE blinkit;

SELECT *
FROM customers as c 
LEFT JOIN orders as o 
ON c.id = o.customer_id
````

## LEFT JOIN with GROUP BY

````python
%%sql

SELECT
    CONCAT(c.first_name, ' ', c.last_name) as customer_name,
    CONCAT('$' ,SUM(o.amount)) as total_spent
FROM customers as c 
LEFT JOIN orders as o 
ON c.id = o.customer_id
GROUP BY customer_name
ORDER BY total_spent DESC
````

#### Find the customers who didn't order anything

````python
%%sql

SELECT
    CONCAT(c.first_name, ' ', c.last_name) as customer_name,
    CONCAT('$' ,SUM(o.amount)) as total_spent
FROM customers as c 
LEFT JOIN orders as o 
ON c.id = o.customer_id
GROUP BY customer_name
HAVING total_spent IS NULL
ORDER BY total_spent DESC
````

### Using IFNULL
```sql
IFNULL(expression, 0)
```

````python
%%sql

SELECT
    CONCAT(c.first_name, ' ', c.last_name) as customer_name,
    IFNULL(SUM(o.amount), 0) as total_spent
FROM customers as c 
LEFT JOIN orders as o 
ON c.id = o.customer_id
GROUP BY customer_name
ORDER BY total_spent DESC
````
