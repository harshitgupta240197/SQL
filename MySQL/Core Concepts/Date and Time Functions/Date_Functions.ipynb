# Date Functions

#### DAY() - Extract the Day of the month from DATES

```python
%%sql

USE dog_world;

SELECT 
    name,
    birth_date,
    DAY(birth_date)
FROM people
```

#### DAYOFWEEK() - Using this we can extract the Day of the week from the DATE

```python
%%sql

USE dog_world;

SELECT 
    name,
    birth_date,
    DAY(birth_date) AS Birth_Date, 
    DAYOFWEEK(birth_date) AS WeekDay
FROM people
```

#### DAYOFYEAR() - Using this we can extract the day of YEAR

```python
%%sql

USE dog_world;

SELECT 
    name,
    birth_date,
    DAY(birth_date) AS Birth_Date, 
    DAYOFWEEK(birth_date) AS WeekDay,
    DAYOFYEAR(birth_date) AS Dayofyear
FROM people
```

#### MONTHNAME()

```python
%%sql

USE dog_world;

SELECT 
    name,
    birth_date,
    DAY(birth_date) AS Birth_Date, 
    DAYOFWEEK(birth_date) AS WeekDay,
    DAYOFYEAR(birth_date) AS Dayofyear,
    MONTHNAME(birth_date) AS MonthName
FROM people
```

#### YEAR()

```python
%%sql

USE dog_world;

SELECT 
    name,
    birth_date,
    DAY(birth_date) AS Birth_Date, 
    DAYOFWEEK(birth_date) AS WeekDay,
    DAYOFYEAR(birth_date) AS Dayofyear,
    MONTHNAME(birth_date) AS MonthName,
    YEAR(birth_date)
FROM people
```
