# ORDER BY

```sql
SELECT 
    column1,
    column2,
    column3
FROM table_name
WHERE condition
GROUP BY column2
ORDER BY column1
```
#### ASC and DESC order can be controlled but ASC is by default

````python
%%sql

USE book_shop;

SELECT 
    book_id,
    author_fname,
    author_lname
FROM books
ORDER BY author_lname DESC
````

````python
%%sql

SELECT
    book_id,
    author_fname,
    author_lname
FROM books
ORDER BY book_id DESC
````

````python
%%sql

SELECT
    book_id,
    author_fname,
    author_lname,
    pages
FROM books
ORDER BY pages DESC
````

````python
%%sql

SELECT
    book_id,
    author_fname,
    author_lname,
    released_year
FROM books
ORDER BY released_year DESC
````

# ORDER BY - shorthand syntax 
```sql
SELECT 
    column1,
    column2,
    column3
FROM table_name
WHERE condition
ORDER BY 2
```
#### Here 2 is basically the second column in the SELECT statement

````python
%%sql

SELECT 
    title,
    author_fname, 
    author_lname
FROM books
ORDER BY 2;
````

# MULTI COLUMN SORT BY
```sql
SELECT 
    column1,
    column2,
    column3
FROM table_name
WHERE condition
GROUP BY column2
ORDER BY column1, column2
```
#### So in such cases the ORDER BT takes place for the first column and then moves to the ORDER BY for the next column

````python
%%sql

SELECT 
    book_id,
    title,
    author_fname,
    released_year
FROM books
ORDER BY released_year, book_id
````

# We can do ORDER BY also via alias

````python
%%sql

SELECT 
    CONCAT(author_fname, ' ', author_lname) AS author
FROM books
ORDER BY author
````
