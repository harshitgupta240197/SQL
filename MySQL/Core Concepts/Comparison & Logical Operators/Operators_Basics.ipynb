#### Connection to MySQL

```python
%load_ext sql
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False

import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

load_dotenv()

url = URL.create("mysql+pymysql",
                 username="root",
                 password=os.getenv('MYSQL_PW'),
                 host="localhost")
engine = create_engine(url)
%sql engine
```

# Not Equal (!=)

```python
%%sql

USE book_shop;

SELECT *
FROM books
WHERE released_year != 2003
```

```python
%%sql

SELECT *
FROM books
WHERE author_lname != 'Harris'
```

# NOT LIKE

##### I want to find all the books that do not have a space in their title

```python
%%sql

SELECT
    title
FROM books
WHERE title NOT LIKE '% %'
```

```python
%%sql

SELECT 
    author_fname
FROM books
WHERE author_fname NOT LIKE '%Da%'
```

# Greater than (>)

```python
%%sql

SELECT *
FROM books
WHERE released_year > 2010
```

# Less Than or Equal To

```python
%%sql

SELECT *
FROM books
WHERE released_year <= 2000
```

# Logical AND (&&)

#### Find all the books written by Dave and published after 2010

```python
%%sql

SELECT *
FROM books
WHERE
    author_fname = 'Dave' &&
    released_year > 2010
```

#### Find the books that have a title greater than 15 and has many pages than all

```python
%%sql

SELECT *
FROM books
WHERE
    CHAR_LENGTH(title) > 15 AND
    pages > 500
```

# Logical OR

```python
%%sql

SELECT *
FROM books
WHERE
    author_fname = 'Dave' OR
    released_year > 2010
```

# BETWEEN

#### Get all the books published between 2004 and 2015

```python
%%sql

SELECT *
FROM books
WHERE released_year BETWEEN 2004 AND 2015
```

# NOT BETWEEN

```python
%%sql

SELECT *
FROM books
WHERE released_year NOT BETWEEN 2004 AND 2015
```

# Comparing Dates

```python
%%sql

USE dog_world;

SELECT *
FROM people
WHERE birth_date < '2005-01-01'
```

```python
%%sql

SELECT *
FROM people
WHERE YEAR(birth_date) < 2005
```

# CAST

```python
%%sql

SELECT 
    CAST('09:00:00' AS TIME)
```

```python
%%sql

SELECT *
FROM people
WHERE birth_time 
    BETWEEN 
        CAST('10:00:00' AS TIME) AND 
        CAST('12:00:00' AS TIME)
```

# IN Operator

```python
%%sql

USE book_shop;

SELECT *
FROM books
WHERE author_lname IN ('Carver', 'Lahiri', 'Smith')
```

# NOT IN Operator

```python
%%sql

USE book_shop;

SELECT *
FROM books
WHERE author_lname NOT IN ('Carver', 'Lahiri', 'Smith', 'Harris', 'Dave')
```

```python
%%sql

SELECT *
FROM books
WHERE released_year % 2 = 0
```
