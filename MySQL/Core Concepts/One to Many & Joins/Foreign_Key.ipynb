# Working with Foreign Key

```python
%%sql

CREATE DATABASE blinkit
```

```python
%%sql

USE blinkit;

CREATE TABLE customers
    (
        id INT AUTO_INCREMENT PRIMARY KEY,
        first_name VARCHAR(50),
        last_name VARCHAR(50),
        email VARCHAR(50)
    );
```

## FOREIGN KEY
#### Adding the FOREIGN KEY constraint makes sure that nobody can add a customer_id in the orders table that does not matches the id column of the customers table

```python
%%sql

CREATE TABLE orders
    (
        id INT AUTO_INCREMENT PRIMARY KEY,
        order_date DATE,
        amount DECIMAL(8,2),
        customer_id INT,
        FOREIGN KEY (customer_id) REFERENCES customers(id)
    );
```
