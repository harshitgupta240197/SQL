# INNER JOIN / JOIN
#### INNER JOIN in SQL selects rows from multiple tables only if they have matching values in the specified shared columns. It acts as a data filter, completely throwing away non-matching rows from both tables.

```python
%%sql

USE blinkit;

SELECT *
FROM customers as c 
INNER JOIN orders as o 
ON c.id = o.customer_id
```
