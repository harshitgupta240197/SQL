# FIRST_VALUE()
#### FIRST_VALUE() is a window function introduced in MySQL 8.0 that returns the value of a specified column or expression from the first row in a defined window frame or partition

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    FIRST_VALUE(emp_no) OVER(PARTITION BY department ORDER BY salary DESC) AS Highest_paid_emp_dept,
    FIRST_VALUE(emp_no) OVER(ORDER BY salary DESC) AS highest_paid_emp_overall
FROM employees
```
