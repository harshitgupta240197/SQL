#### 1. Create a new database 'shirts_db'

```python
%%sql

CREATE DATABASE shirts_db
```

#### 2. Create a new table called 'shirts'
#### 3. Columns : shirt_id, article, color, shirt_size, last_worn 
#### 4. Primary Key is shirt_id

```python
%%sql

USE shirts_db;

CREATE TABLE shirts
    (
        shirt_id INT AUTO_INCREMENT,
        article VARCHAR(50) NOT NULL,
        color VARCHAR(50) NOT NULL,
        shirt_size VARCHAR(10) NOT NULL,
        last_worn INT NOT NULL,
        PRIMARY KEY(shirt_id)
    )
```

#### 5. Insert provided data

```python
%%sql

INSERT INTO shirts (article, color, shirt_size, last_worn)
VALUES
    ('t-shirt', 'white', 'S', 10),
    ('t-shirt', 'green', 'S', 200),
    ('polo shirt', 'black', 'M', 10),
    ('tank top', 'blue', 'S', 50),
    ('t-shirt', 'pink', 'S', 0),
    ('polo shirt', 'red', 'M', 5),
    ('tank top', 'white', 'S', 200),
    ('tank top', 'blue', 'M', 15)
```

#### 6. Add a New Shirt : Purple polo shirt, size M, last worn 50 days ago

```python
%%sql

INSERT INTO shirts (article, color, shirt_size, last_worn)
VALUES ('polo shirt', 'purple', 'M', 50)
```

```python
%%sql

SELECT *
FROM shirts
```

#### 7. Select all shirts but only print out article and color

```python
%%sql

SELECT 
    article,
    color
FROM shirts
```

#### 8. Select all medium shirts. Print out everything BUT shirt_id

```python
%%sql

SELECT 
    article,
    color,
    shirt_size,
    last_worn
FROM shirts
WHERE shirt_size = 'M'
```

#### 9. Update all polo shirts : Change theur size to L

```python
%%sql

UPDATE shirts SET shirt_size = 'L'
WHERE article = 'polo shirt'
```

```python
%%sql

SELECT *
FROM shirts
```

#### 10. Update the shirt that was last worn 15 days ago -> Change last_worn to 0

```python
%%sql

UPDATE shirts SET last_worn = 0
WHERE last_worn = 15
```

```python
%%sql

SELECT *
FROM shirts
```

#### 11. Update all white shirts -> Change size to 'XS' and color to 'off white'

```python
%%sql

UPDATE shirts SET shirt_size = 'XS', color = 'off white'
WHERE color = 'white'
```

```python
%%sql

SELECT *
FROM shirts
```

#### 12. Delete all old shirts -> Last worn 200 days ago

```python
%%sql

DELETE FROM shirts
WHERE last_worn = 200
```

```python
%%sql

SELECT *
FROM shirts
```

#### 13. Delete all tank tops

```python
%%sql

DELETE FROM shirts
WHERE article = 'tank top'
```

```python
%%sql

SELECT *
FROM shirts
```

#### 14. Delete all shirts

```python
%%sql

DELETE FROM shirts
```

#### 15. Drop the enture shirts table

```python
%%sql

DROP TABLE shirts
```
