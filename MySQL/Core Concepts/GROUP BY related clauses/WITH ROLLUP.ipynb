# WITH ROLLUP
#### WITH ROLLUP is a modifier added to the GROUP BY clause that automatically calculates hierarchical subtotals and a grand total within a single query result.

```python
%%sql

USE tv_shows;

SELECT 
    AVG(rating)
FROM full_review
```

```python
%%sql

SELECT 
    title,
    AVG(rating) AS avg_rating
FROM full_review
GROUP BY title
```

# WITH ROLLUP
#### This will add a NULL/NONE row at the end which will have the summary of the entire table

```python
%%sql

SELECT 
    title,
    AVG(rating) AS avg_rating
FROM full_review
GROUP BY title
WITH ROLLUP
```

```python
%%sql

SELECT 
    title,
    COUNT(rating) AS count_rating
FROM full_review
GROUP BY title
WITH ROLLUP
```

```python
%%sql

SELECT 
    title,
    genre,
    COUNT(rating) AS count_rating
FROM full_review
GROUP BY title, genre
WITH ROLLUP
```
