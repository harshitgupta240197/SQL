# STRICT TRANS TABLES (TRANSACTIONAL)

####  is a MySQL mode that forces the database to reject bad data instead of changing it quietly. If you try to save a wrong value, the server stops the command and shows an error.
## What It Does

- Stops bad inserts when text is too long or numbers are out of range.
- Rejects `NULL` values sent to columns that do not allow empty data.
- Stops rollbacks or cancels full transactions on transactional engines like InnoDB if a row fails.
- Prevents hidden data corruption from silent truncation.
