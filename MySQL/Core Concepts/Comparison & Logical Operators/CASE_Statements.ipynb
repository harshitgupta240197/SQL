#### Connection to MySQL

```python
%load_ext sql
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False

import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

load_dotenv()

url = URL.create("mysql+pymysql",
                 username="root",
                 password=os.getenv('MYSQL_PW'),
                 host="localhost")
engine = create_engine(url)
%sql engine
```

# CASE Statements

```python
%%sql

USE book_shop;

SELECT
    title, 
    released_year,
    CASE
        WHEN released_year BETWEEN 2001 AND 2012 THEN 'Modern Literature'
        WHEN released_year > 2012 THEN 'Genz Literature'
        ELSE '20th Cent Lit'
    END AS Genre
FROM books
```

```python
%%sql

SELECT
    title,
    stock_quantity,
    CASE
        WHEN stock_quantity BETWEEN 0 AND 10 THEN '*'
        WHEN stock_quantity BETWEEN 11 AND 20 THEN '**'
        WHEN stock_quantity BETWEEN 21 AND 30 THEN '***'
        ELSE '****'
    END AS stock_rating
FROM books
```
