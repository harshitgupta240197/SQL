#### Connection to MySQL

````python
%load_ext sql
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False

import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

load_dotenv()

url = URL.create("mysql+pymysql",
                 username="root",
                 password=os.getenv('MYSQL_PW'),
                 host="localhost")
engine = create_engine(url)
%sql engine
````

# LIKE
```sql
SELECT
    column1,
    column2
FROM table_name
WHERE column1 LIKE '%search-term%'
```
#### Here % is the wildcard

````python
%%sql

USE book_shop;

SELECT *
FROM books
WHERE author_fname LIKE '%Dave%'
LIMIT 5;
````

# LIKE - UNDERSCORES(_) 
#### Underscore _ represents a character

#### We will use this LIKE _ to find the authors with first name as exactly 4 characters

````python
%%sql

SELECT *
FROM books
WHERE author_fname LIKE '____'
````

````python
%%sql

SELECT *
FROM books
WHERE author_fname LIKE '_a__'
````
