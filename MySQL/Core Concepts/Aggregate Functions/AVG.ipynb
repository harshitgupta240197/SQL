# AVG

#### Find the average number of pages for all the books combined in the database

```python
%%sql

USE book_shop;

SELECT 
    AVG(pages)
FROM books
```

#### AVG all the pages each author has written

```python
%%sql

SELECT
    CONCAT(author_fname, ' ', author_lname) AS author_name,
    AVG(pages) AS average_pages
FROM books
GROUP BY author_name
ORDER BY average_pages DESC
```

#### Calculate the average stock quantity for books released in the same year

```python
%%sql

SELECT 
    released_year,
    AVG(stock_quantity) AS avg_stock_quantity,
    COUNT(*) AS '# of books'
FROM books
GROUP BY released_year
```
