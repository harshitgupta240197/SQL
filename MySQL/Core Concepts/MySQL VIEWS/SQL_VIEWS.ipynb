# SQL VIEWS
#### SQL view is a virtual table generated by the result set of a pre-defined SELECT query. Unlike standard database tables, a standard view does not store physical data; instead, it acts as a dynamic window that queries underlying "base" tables in real time whenever it is called.

```python
%%sql

USE tv_shows;

SHOW TABLES
```

```python
%%sql

USE tv_shows;

SELECT *
FROM series
LIMIT 5;
```

```python
%%sql

USE tv_shows;

SELECT *
FROM reviews
LIMIT 5;
```

```python
%%sql

USE tv_shows;

SELECT *
FROM reviewers
LIMIT 5;
```

```python
%%sql

CREATE VIEW full_review AS 
SELECT 
    title,
    released_year,
    genre,
    rating,
    first_name,
    last_name
FROM series as s 
JOIN reviews as r
ON s.id = r.series_id
JOIN reviewers as rv 
ON r.reviewer_id = rv.id
```

```python
%%sql

SELECT *
FROM full_review
LIMIT 5;
```

# REPLACE - altering a VIEW
# CREATE OR REPLACE VIEW
# ALTER VIEW

```python
%%sql

CREATE OR REPLACE VIEW full_review AS 
SELECT 
    title,
    released_year,
    genre,
    rating,
    first_name,
    last_name
FROM series as s 
JOIN reviews as r
ON s.id = r.series_id
JOIN reviewers as rv 
ON r.reviewer_id = rv.id
ORDER BY title DESC
```

```python
%%sql

ALTER VIEW full_review AS 
SELECT 
    title,
    released_year,
    genre,
    rating,
    first_name,
    last_name
FROM series as s 
JOIN reviews as r
ON s.id = r.series_id
JOIN reviewers as rv 
ON r.reviewer_id = rv.id
ORDER BY genre DESC
```
