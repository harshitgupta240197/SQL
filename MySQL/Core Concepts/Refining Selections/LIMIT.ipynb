# LIMIT (Selective Range)

```sql
SELECT *
FROM table_name
LIMIT 4, 10
```
#### So here 4 is the point where we want to extract from and 10 defines the number of rows we want
#### Basically 4 is the Starting row and 10 says how many rows to extract
