# SUBSTRING / SUBSTR - String Function
```sql
SELECT
    SUBSTRING(column_1, start_point, end_point)
FROM table_name
```

````python
%%sql

USE book_shop;

SELECT 
    SUBSTRING(title, 1, 15) AS partial_name
FROM books
LIMIT 5;
````

### We will try to get the first character of author's last name

````python
%%sql

SELECT
    SUBSTRING(author_lname, 1, 1) AS author_initials
FROM books
LIMIT 5;
````
