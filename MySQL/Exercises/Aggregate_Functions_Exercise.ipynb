# Aggregate Functions Exercise

```python
%%sql

USE book_shop;

SELECT *
FROM books
LIMIT 2;
```

#### Print the number of books in the database

```python
%%sql

SELECT
    COUNT(*) AS total_books
FROM books
```

#### Print out how many books were released in each year

```python
%%sql

SELECT
    released_year,
    COUNT(*) AS '# of books'
FROM books
GROUP BY released_year
```

#### Print out the total number of books in stock

```python
%%sql

SELECT 
    SUM(stock_quantity)
FROM books
```

#### Find the average released_year for each author

```python
%%sql

SELECT 
    CONCAT(author_fname, ' ', author_lname) AS author_name,
    ROUND(AVG(released_year), 0)
FROM books
GROUP BY author_name
```

#### Find the full name of the author who wrote the longest book

```python
%%sql

SELECT
    CONCAT(author_fname, ' ', author_lname) AS author_name,
    pages
FROM books
WHERE pages = (
    SELECT 
        MAX(pages)
    FROM books
)
```

```python
%%sql

SELECT
    released_year,
    COUNT(*) AS '# books',
    AVG(pages)
FROM books
GROUP BY released_year
ORDER BY released_year
```
