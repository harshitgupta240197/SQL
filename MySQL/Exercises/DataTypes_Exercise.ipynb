#### Create a Tweets table that stores
#### 1. The tweet content 2. A username 3. Time it was created

```python
%%sql

USE dog_world;

CREATE TABLE tweets
    (
        text_content VARCHAR(100),
        username VARCHAR(20), 
        created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
    );
```

```python
%%sql

INSERT INTO tweets (text_content, username)
VALUES
    ('Hey ! Everyone, I am new here!', 'genguy@gmail.com'),
    ('Yo! Just Chillin!', 'menguy@gmail.com'),
    ('I am the owner of this space!', 'heng@gmail.com')
```

```python
%%sql

SELECT *
FROM tweets
```
