# WINDOW FUNCTIONS
#### Window functions in MySQL perform calculations across a set of table rows related to the current row while retaining individual row details instead of collapsing them into a single summary row like a standard GROUP BY clause.
#### Window functions perform aggregate operations on group of rows, but they produce a result for each row

# OVER()
#### The OVER() clause constructs a window. When it is empty, the window will include all records

```python
%%sql

USE dog_world;

CREATE TABLE employees (
    emp_no INT PRIMARY KEY AUTO_INCREMENT,
    department VARCHAR(20),
    salary INT
);
```

#### Using a normal GROUP BY

```python
%%sql

USE dog_world;

SELECT 
    department,
    AVG(salary)
FROM employees
GROUP BY department
```

#### Using OVER()

```python
%%sql

SELECT 
    department,
    AVG(salary) OVER()
FROM employees
LIMIT 5;
```

```python
%%sql

SELECT 
    emp_no,
    department,
    AVG(salary) OVER()
FROM employees
```

```python
%%sql

SELECT 
    emp_no,
    department,
    MIN(salary) OVER(),
    MAX(salary) OVER()
FROM employees
LIMIT 10;
```
