# HAVING clause
#### It is used to filter the groups that we get via GROUP BY

```python
%%sql

USE tv_shows;

SELECT 
    title,
    AVG(rating) as average_rating
FROM full_review
GROUP BY title
HAVING average_rating > 8
```
