# PARTITION BY
### OVER(PARTITION BY)
#### Inside of the OVER(), use PARTITION BY to form rows into groups of row
#### The PARTITION BY clause divides a result set into smaller groups (partitions) so that calculations can be performed independently within each group.

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    AVG(salary) OVER(PARTITION BY department) AS department_avg
FROM employees
```

```python
%%sql

SELECT 
    emp_no, 
    department,
    salary,
    COUNT(*) OVER(PARTITION BY department) AS department_count
FROM employees
```

```python
%%sql

SELECT 
    emp_no, 
    department,
    salary,
    SUM(salary) OVER(PARTITION BY department) AS department_payroll,
    SUM(salary) OVER() AS total_payroll
FROM employees
```
