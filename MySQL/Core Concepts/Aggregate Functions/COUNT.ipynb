#### Connection to MySQL

````python
%load_ext sql
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False

import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

load_dotenv()

url = URL.create("mysql+pymysql",
                 username="root",
                 password=os.getenv('MYSQL_PW'),
                 host="localhost")
engine = create_engine(url)
%sql engine
````

# COUNT

#### Count the number of rows in the books table of the book_shop database

````python
%%sql

USE book_shop;

SELECT 
    COUNT(*)
FROM books
````

#### Count the number of author first names

````python
%%sql

SELECT 
    COUNT(author_fname) AS 'author name count'
FROM books
````

# COUNT DISTINCT
```sql
SELECT
    COUNT(DISTINCT column_name)
FROM table_name
```

````python
%%sql

SELECT 
    COUNT(DISTINCT author_fname)
FROM books
````

````python
%%sql

SELECT 
    COUNT(DISTINCT released_year)
FROM books
````
