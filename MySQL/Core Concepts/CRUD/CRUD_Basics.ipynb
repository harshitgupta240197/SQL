# Basics of CRUD
#### Create, Read, Update, and Delete

# CREATE

````python
%%sql

USE dog_world;

CREATE TABLE cats (
    cat_id INT AUTO_INCREMENT,
    name VARCHAR(100),
    breed VARCHAR(100),
    age INT,
    PRIMARY KEY (cat_id)
); 
````

````python
%%sql

INSERT INTO cats(name, breed, age) 
VALUES ('Ringo', 'Tabby', 4),
       ('Cindy', 'Maine Coon', 10),
       ('Dumbledore', 'Maine Coon', 11),
       ('Egg', 'Persian', 4),
       ('Misty', 'Tabby', 13),
       ('George Michael', 'Ragdoll', 9),
       ('Jackson', 'Sphynx', 7);
````

# READ

````python
%%sql

SELECT *
FROM cats 
````

````python
%%sql

SELECT
    name,
    breed
FROM cats 
````

````python
%%sql

SELECT *
FROM cats 
WHERE name = 'Egg'
````

````python
%%sql

SELECT *
FROM cats 
WHERE age > 4
````

# Update
#### Always do a SELECT query to confirm before UPDATE to make sure your target is accurate 
```sql
UPDATE table_name SET column_name = 'updated_value'
WHERE condition
```

````python
%%sql

USE dog_world;

CREATE TABLE employees
    (
        employee_id INT AUTO_INCREMENT,
        first_name VARCHAR(50) NOT NULL,
        last_name VARCHAR(50) NOT NULL,
        position VARCHAR(50) NOT NULL,
        PRIMARY KEY(employee_id)
    )
````

````python
%%sql

INSERT INTO employees (first_name, last_name, position)
VALUES
    ('Adam', 'Shaw', ' Lead Engineer'),
    ('Darin', 'John', ' Junior Engineer'),
    ('Stephen', 'Kyler', ' BI Engineer'),
    ('Kayla', 'Johnson', ' Lead BI Engineer'),
    ('Brad', 'Tyler', ' Senior Software Engineer');
````

````python
%%sql

SELECT *
FROM employees
````

#### UPDATE

````python
%%sql

UPDATE employees SET last_name = 'Johnson'
WHERE 
    first_name = 'Darin';
````

````python
%%sql

SELECT *
FROM employees
````

#### Multi - UPDATE

````python
%%sql

UPDATE employees SET last_name = 'bad_johnson', position = 'termianted'
WHERE first_name = 'Darin'
````

````python
%%sql

SELECT *
FROM employees
````
