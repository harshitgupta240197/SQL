# Auto-Increment
```sql
CREATE TABLE table_name
    (
        id_column INT AUTO_INCREMENT,
        column_1 VARCHAR(100), 
        column_2 INT,
        PRIMARY KEY(id_column)
    )
```

````python
%%sql

USE dog_world;

CREATE TABLE dogs_5
    (
        dog_id INT AUTO_INCREMENT,
        name VARCHAR(100) NOT NULL,
        age INT NOT NULL,
        PRIMARY KEY(dog_id)
    )
````

````python
%%sql

INSERT INTO dogs_5 (name, age)
VALUES
    ('ginger', 2),
    ('scatt', 7),
    ('bratt', 4),
    ('flat', 1),
    ('matt', 2);
````

#### We can see that the dog_id got filled through AUTO_INCREMENT

````python
%%sql

SELECT *
FROM dogs_5
````
