# Comparison & Logical Operators Exercise

#### Find all the books written before 1980

```python
%%sql

USE book_shop;

SELECT *
FROM books
WHERE released_year < 1980
```

#### Find all the books written by Eggers or Chabon

```python
%%sql

SELECT *
FROM books
WHERE author_lname IN ('Eggers', 'Chabon')
```

#### Select all the books written by Lahiri and published after 2000

```python
%%sql

SELECT *
FROM books
WHERE 
    author_lname = 'Lahiri' AND
    released_year > 2000
```

#### Find all the books with page counts between 100 and 200

```python
%%sql

SELECT *
FROM books
WHERE pages BETWEEN 100 AND 200
```

#### Select all books where author_lname starts with a C or S

```python
%%sql

SELECT *
FROM books
WHERE 
    author_lname LIKE 'C%' OR
    author_lname LIKE 'S%'
```

```python
%%sql

SELECT
    title,
    author_lname, 
    CASE
        WHEN title LIKE '%stories%' THEN 'Short Stories'
        WHEN title LIKE '%Just Kids%' THEN 'Memoir'
        WHEN title LIKE '%Heartbreaking Work%' THEN 'Memoir'
        ELSE 'Novel'
    END AS TYPE
FROM books
        
```

```python
%%sql

SELECT 
    author_fname,
    author_lname,
    CONCAT(COUNT(*), (
            CASE
                WHEN COUNT(*) > 1 THEN ' books'
                ELSE 'book'
                END
    )) AS 'COUNT'
FROM books
GROUP BY author_fname, author_lname
```
