# REPLACE - String Function
```sql
SELECT
    REPLACE('Target', 'WhatToReplace', 'WhatToReplaceWith')
FROM table_name
```
### It is a case sensitive function


#### As a task replace all the white spaces in book title with hyphen

````python
%%sql

USE book_shop;

SELECT 
    REPLACE(title, ' ', '-') AS new_title
FROM books
LIMIT 5;
````
