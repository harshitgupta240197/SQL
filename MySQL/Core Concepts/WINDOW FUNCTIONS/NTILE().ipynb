# NTILE()
#### The NTILE(N) window function in MySQL divides an ordered result set into N approximately equal-sized groups (buckets) and assigns a bucket number (from 1 to N) to each row. It is commonly used for statistical calculations like quartiles (NTILE(4)), deciles (NTILE(10)), or percentiles (NTILE(100))

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    NTILE(4) OVER(ORDER BY salary DESC) AS salary_quartile
FROM employees
```

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    NTILE(4) OVER(PARTITION BY department ORDER BY salary DESC) AS dept_salary_quart,
    NTILE(4) OVER(ORDER BY salary DESC) AS salary_quartile
FROM employees
```
