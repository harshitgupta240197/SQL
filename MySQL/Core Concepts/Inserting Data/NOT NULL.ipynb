# NOT NULL
#### NULL and 0 are not the same since 0 is considered as a value but NULL means there is nothing there (empty).
#### Adding NOT NULL to column as a constraint makes sure the column will never have any NULL value

```sql
CREATE TABLE table_name
    (
        column_1 VARCHAR(100) NOT NULL
        column_2 INT NOT NULL
    );

```

````python
%%sql

USE dog_world;

CREATE TABLE dogs_2
    (
        name VARCHAR(50) NOT NULL,
        age INT NOT NULL
    );
````

#### When we do a DESCRIBE on this new table we can clearly see that for NULL column it says NO

````python
%%sql

USE dog_world;

DESCRIBE dogs_2
````

````python
%%sql

USE dog_world;

DROP TABLE dogs_2
````
