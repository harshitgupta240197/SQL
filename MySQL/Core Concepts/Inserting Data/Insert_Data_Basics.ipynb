````python
%%sql

USE dog_world;

CREATE TABLE dogs(
    name VARCHAR(50),
    age INT
)
````

#### Create a table called 'people'

#### After creating these columns insert data in the table

````python
%%sql

USE dog_world;

CREATE TABLE people(
    first_name VARCHAR(20),
    last_name VARCHAR(20),
    age INT
);
````

````python
%%sql

USE dog_world;

INSERT INTO people (first_name, last_name, age)
VALUES
    ('Peter', 'Parker', 28),
    ('Tom', 'Holland', 32),
    ('Ben', 'Affleck', 72),
    ('Tony', 'Irons', 80)
````

````python
%%sql

USE dog_world;

SELECT *
FROM people
````
