# Exercise on String Functions

### 1. Reverse and Upper case the following sentence

##### 'Why does my dog keep on eating icecream?'

```python
%%sql

SELECT 
    UPPER(REVERSE('Why does my dog keep on eating icecream?'))
```

```python
%%sql

SELECT
  REPLACE
  (
  CONCAT('I', ' ', 'like', ' ', 'cats'),
  ' ',
  '_'
  );
```

```python
%%sql

USE book_shop;

SELECT 
    REPLACE(title, ' ', '->')
FROM books
```

```python
%%sql

SELECT 
    author_fname AS forwards,
    REVERSE(author_lname) AS backwards
FROM books
```

```python
%%sql

SELECT 
    UPPER(
        CONCAT(
            author_fname,
            ' ',
            author_lname
            )) 
            AS 'full name in caps'
FROM books
```

```python
%%sql

SELECT 
    CONCAT(title, ' was released in ', released_year) AS blurb
FROM books
```

```python
%%sql

SELECT
    title, 
    CHAR_LENGTH(title) AS 'character count'
FROM books
LIMIT 5;
```

```python
%%sql

SELECT 
    CONCAT(SUBSTRING(title, 1, 10), '...') AS 'short title', 
    CONCAT(author_lname, ' ', author_fname) AS author,
    CONCAT(stock_quantity, ' ', 'in stock') AS quantity
FROM books
```
