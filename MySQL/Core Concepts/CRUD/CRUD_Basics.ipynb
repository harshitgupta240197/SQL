#### Connection to MySQL

```python
%load_ext sql
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False

import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

load_dotenv()

url = URL.create("mysql+pymysql",
                 username="root",
                 password=os.getenv('MYSQL_PW'),
                 host="localhost")
engine = create_engine(url)
%sql engine
```

# Basics of CRUD
#### Create, Read, Update, and Delete

# CREATE

```python
%%sql

USE dog_world;

CREATE TABLE cats (
    cat_id INT AUTO_INCREMENT,
    name VARCHAR(100),
    breed VARCHAR(100),
    age INT,
    PRIMARY KEY (cat_id)
); 
```

```python
%%sql

INSERT INTO cats(name, breed, age) 
VALUES ('Ringo', 'Tabby', 4),
       ('Cindy', 'Maine Coon', 10),
       ('Dumbledore', 'Maine Coon', 11),
       ('Egg', 'Persian', 4),
       ('Misty', 'Tabby', 13),
       ('George Michael', 'Ragdoll', 9),
       ('Jackson', 'Sphynx', 7);
```

# READ

```python
%%sql

SELECT *
FROM cats 
```

```python
%%sql

SELECT
    name,
    breed
FROM cats 
```

```python
%%sql

SELECT *
FROM cats 
WHERE name = 'Egg'
```

```python
%%sql

SELECT *
FROM cats 
WHERE age > 4
```
