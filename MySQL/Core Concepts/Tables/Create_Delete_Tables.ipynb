````python
%load_ext sql
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False

import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

load_dotenv()

url = URL.create("mysql+pymysql",
                 username="root",
                 password=os.getenv('MYSQL_PW'),
                 host="localhost")
engine = create_engine(url)
%sql engine
````

# Creating Tables
```sql
CREATE TABLE table_name 
    (
        column_1 VARCHAR(100),
        column_2 INT
    );
```

#### Let us create a sample database called 'dog_world'

````python
%%sql

CREATE DATABASE dog_world
````

#### Inside dog_world database we will create a table called 'dogs'
#### For the name column we will give a VARCHAR limit of 50

````python
%%sql

USE dog_world;

CREATE TABLE dogs
(
    name VARCHAR(50),
    age INT
);
````

# Show Tables in the current Database

````python
%%sql

USE dog_world;

SHOW TABLES
````

# Show Columns in a particular Table

````python
%%sql

USE dog_world;

SHOW COLUMNS FROM dogs
````

# DESCRIBE / DESC a Table

````python
%%sql

USE dog_world;

DESCRIBE dogs
````

# Dropping/Deleting Tables

````python
%%sql

USE dog_world;

DROP TABLE dogs;
````

````python
%%sql

SHOW TABLES
````
