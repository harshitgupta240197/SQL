# DEFAULT Values
### We can assign a default value to a column in a table so that if nothing is entered then it shows that default value
```sql
CREATE TABLE table_name
    (
        column_1 VARCHAR(100) DEFAULT 'default_value',
        column_2 INT DEFAULT 51
    );
```

````python
%%sql

USE dog_world;

CREATE TABLE dogs_3
    (
        name VARCHAR(100) DEFAULT 'john_doe',
        age INT DEFAULT 7
    );
````

````python
%%sql

USE dog_world;

INSERT INTO dogs_3 (name, age)
VALUES
    ('jack', 9),
    ('mary', 2),
    (DEFAULT, 5),
    ('leslie', 6)
````

````python
%%sql

SELECT *
FROM dogs_3
````

### NOT NULL and DEFAULT are not the same. You can set a value to NULL if you just have a DEFAULT value and NOT NULL is not enabled for that column

````python
%%sql

INSERT INTO dogs_3 (name, age)
VALUES
    ('jack', 12),
    ('ginger', 10),
    ('salmon', 8),
    (NULL, NULL)
````

````python
%%sql

SELECT *
FROM dogs_3
````

````python
%%sql

DROP TABLE dogs_3
````
