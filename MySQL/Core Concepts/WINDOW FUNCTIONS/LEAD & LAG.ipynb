# LEAD & LAG
#### LEAD() and LAG() are window functions used to look forward or backward into adjacent rows within a result set without needing a self-join

# LAG()

#### If we apply LAG it will evalaute the previous employee salary

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    LAG(salary) OVER(ORDER BY salary DESC) AS lag_salary
FROM employees
```

### Calcualte the salary diff

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    salary - LAG(salary) OVER(ORDER BY salary DESC) AS salary_diff
FROM employees
```

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    salary - LAG(salary) OVER(PARTITION BY department ORDER BY salary DESC) AS dept_salary_diff,
    salary - LAG(salary) OVER(ORDER BY salary DESC) AS salary_diff
FROM employees
```

# LEAD()

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    LEAD(salary) OVER(ORDER BY salary DESC) AS lead_salary
FROM employees
```

```python
%%sql

USE dog_world;

SELECT 
    emp_no,
    department,
    salary,
    salary - LEAD(salary) OVER(ORDER BY salary DESC) AS salary_diff
FROM employees
```
