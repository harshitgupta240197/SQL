```python
%load_ext sql
%config SqlMagic.autopandas = True
%config SqlMagic.feedback = False

import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

load_dotenv()

url = URL.create("mysql+pymysql",
                 username="root",
                 password=os.getenv('MYSQL_PW'),
                 host="localhost")
engine = create_engine(url)
%sql engine
```

```python
%%sql

SELECT
    id,
    username,
    created_at
FROM users
ORDER BY created_at
LIMIT 5;

```

```python
%%sql

SELECT
    DAYNAME(created_at),
    COUNT(id)
FROM users
GROUP BY DAYNAME(created_at)
ORDER BY COUNT(id) DESC
```

```python
%%sql

SELECT *
FROM users as u 
LEFT JOIN photos as p 
ON u.id = p.user_id
WHERE p.user_id IS NULL
```

```python
%%sql

SELECT 
    u.id,
    p.id as photo_id,
    u.username,
    p.image_url,
    COUNT(*)
FROM users as u 
JOIN photos as p 
ON u.id = p.user_id
JOIN likes as l 
ON p.id = l.photo_id
GROUP BY p.id
ORDER BY COUNT(*) DESC
```

```python
%%sql

SELECT 
    SUM(p.id) / SUM(u.id) AS avg_post_times
FROM users as u 
JOIN photos as p 
ON u.id = p.user_id
```

```python
%%sql

SELECT
    t.tag_name,
    p.tag_id,
    COUNT(p.photo_id)
FROM photo_tags as p
JOIN tags as t 
ON p.tag_id = t.id
GROUP BY p.tag_id
ORDER BY COUNT(p.photo_id) DESC
LIMIT 5;
```

```python
%%sql

SELECT 
    u.username,
    l.user_id,
    COUNT(*) AS num_likes
FROM users as u 
JOIN likes as l 
ON u.id = l.user_id
GROUP BY l.user_id
HAVING num_likes = (
    SELECT
        COUNT(*)
    FROM photos
)

```
