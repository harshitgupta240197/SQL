# MIN(), MAX(), GROUP BY

#### Find the year each author published their first book

```python
%%sql

USE book_shop;

SELECT
    author_fname,
    author_lname,
    COUNT(*) AS books_written,
    MIN(released_year) AS oldest_release,
    MAX(released_year) AS latest_release
FROM books
GROUP BY author_fname, author_lname
```
