# DATE MATH

# DATEDIFF()
```sql
SELECT
    DATEDIFF('first_date', 'second_date')
FROM table_name
```

````python
%%sql

USE dog_world;

SELECT 
    DATEDIFF(
        (
            SELECT (birth_date)
            FROM people
            WHERE name = 'Lulu'
        ),
        (
            SELECT (birth_Date)
            FROM people
            WHERE name = 'Matt'
        )
    )
````

#### Calculating the age

````python
%%sql

SELECT 
    name,
    birth_date,
    ROUND(DATEDIFF(CURDATE(), birth_date) / 365) AS Age
FROM people
````

# DATE_ADD() / DATE_SUB()
```sql
SELECT
    DATE_ADD('your_date', INTERVAL 1 DAY )
FROM table_name
```
________
```sql
SELECT
    DATE_SUB('your_date', INTERVAL 1 YEAR )
FROM table_name
```

````python
%%sql

SELECT 
    name, 
    birth_date,
    DATE_ADD(birth_date, INTERVAL 1 YEAR) AS add_1_year
FROM people
````

#### Check when each of them turned or will turn 18 years of age

````python
%%sql

SELECT 
    name,
    birth_date,
    DATE_ADD(birth_date, INTERVAL 18 YEAR) AS 'will reach 18'
FROM people
````

# TIMEDIFF

````python
%%sql

SELECT 
    name,
    birth_time,
    TIMEDIFF(CURTIME(), birth_time)
FROM people
````
