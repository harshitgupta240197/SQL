# UPPER() & LOWER() case
#### UPPER
```sql
SELECT 
    UPPER(column_name)
FROM table_name
```
#### LOWER
```sql
SELECT 
    LOWER(column_name)
FROM table_name
```

````python
%%sql

USE book_shop;

SELECT *
FROM books
LIMIT 5;
````

````python
%%sql

SELECT
    UPPER(author_fname) AS AUTHOR_FIRST_NAME,
    LOWER(author_lname) AS AUTHOR_LAST_NAME
FROM books
LIMIT 5;
````

````python
%%sql

SELECT 
    CONCAT(
        'I LOVE',
        ' ',
        UPPER(title)
        )
FROM books
LIMIT 5;
````
