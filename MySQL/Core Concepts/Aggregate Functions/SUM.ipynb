# SUM

#### Find the total number of pages for all the books combined in the database

```python
%%sql

USE book_shop;

SELECT 
    SUM(pages)
FROM books
```

#### Sum all the pages each author has written

```python
%%sql

SELECT
    CONCAT(author_fname, ' ', author_lname) AS author_name,
    SUM(pages) AS total_pages
FROM books
GROUP BY author_name
ORDER BY total_pages DESC
```
