# Primary Keys : Unique Identifier
```sql
CREATE TABLE table_name
    (
        id_column INT NOT NULL PRIMARY KEY,
        column_1 VARCHAR(100),
        column_2 INT
    )
```
### Another way
```sql
CREATE TABLE table_name
    (
        id_column INT NOT NULL,
        name VARCHAR(100),
        age INT,
        PRIMARY KEY(id_column)
    );
```

### Specifying NOT NULL for a primary key column is redundant since a primary key column can never be NULL

````python
%%sql

USE dog_world;

CREATE TABLE dogs_4
    (
        dog_id INT NOT NULL PRIMARY KEY,
        name VARCHAR(100) NOT NULL,
        age INT NOT NULL
    )
````

````python
%%sql

INSERT INTO dogs_4 (dog_id, name, age)
VALUES
    (1, 'tom', 5),
    (2, 'yanky', 7),
    (3, 'slue', 2),
    (4, 'green', 3),
    (5, 'gray', 2);
````

### When we do a DESCRIBE on this table we can see that the KEY column shows dog_id as PRIMARY KEY

````python
%%sql

DESCRIBE dogs_4
````

````python
%%sql
SELECT *
FROM dogs_4
````
